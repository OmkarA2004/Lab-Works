```python
from Bio import Entrez
```

```python
Entrez.email = "omkarda.bt24@rvce.edu.in"
Entrez.tool = "MRNA_retriever"

search_query = "human[organism] AND PLAT[gene] AND mRNA[Title]"
```

```python
handle = Entrez.esearch(db="nucleotide", term=search_query)

record = Entrez.read(handle)

handle.close()

id_list = record["IdList"]

count = record["Count"]

print(f"Found {count} records.")

print("First 10 IDs:", id_list[:10])
```

```python
handle = Entrez.efetch(
    db="nucleotide",
    id=id_list,
    rettype="gb",
    retmode="text"
)

mrna_records = handle.read()

handle.close()

print(mrna_records[:500])
```

```python
from Bio import SeqIO

from io import StringIO

mrna_info = []

for record in SeqIO.parse(StringIO(mrna_records), "genbank"):

    mrna_info.append({

        "accession": record.id,

        "description": record.description,

        "sequence": str(record.seq)

    })

print(f"Extracted information for {len(mrna_info)} records.")

if mrna_info:

    print("First record:", mrna_info[0])
```

```python
for record in mrna_info:

    print("Accession:", record["accession"])

    print("Description:", record["description"])

    print("Sequence (first 100 chars):", record["sequence"][:100])

    print("-" * 20)
```
