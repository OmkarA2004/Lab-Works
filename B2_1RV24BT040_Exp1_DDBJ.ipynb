```python
import requests
from Bio import SeqIO
from io import StringIO
```

```python
search_query = 'TPA AND "Homo sapiens" AND mRNA'

print("Search query:", search_query)
```

```python
arsa_url = "https://ddbj.nig.ac.jp/arsa/search"

params = {
    "query": search_query
}

response = requests.get(arsa_url, params=params)

print("Status code:", response.status_code)
print(response.url)
```

```python
accession = "BR000690"

print("Accession:", accession)
```

```python
url = "https://getentry.ddbj.nig.ac.jp/getentry"

params = {
    "database": "na",
    "accession_number": accession,
    "format": "fasta"
}

response = requests.get(url, params=params)

print("Status code:", response.status_code)
```

```python
fasta_data = response.text

print(fasta_data)
```

```python
record = SeqIO.read(StringIO(fasta_data), "fasta")

print("Accession:", record.id)
print("Description:", record.description)
print("Sequence length:", len(record.seq))
print("Sequence:")
print(record.seq)
```
