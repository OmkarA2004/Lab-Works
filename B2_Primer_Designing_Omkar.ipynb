```python
uploaded = files.upload()
filename = next(iter(uploaded))

with open(filename) as f:
    lines = f.readlines()

seq = ''.join(lines[1:]).replace('\n','').replace('\r','').upper()

print("Sequence:", filename)
print("Length:", len(seq), "bp")
```

```python
import primer3

seq = seq.strip().upper()

settings = {
    "SEQUENCE_ID": "ACTB",
    "SEQUENCE_TEMPLATE": seq,
    "PRIMER_NUM_RETURN": 10,
    "PRIMER_OPT_SIZE": 20,
    "PRIMER_MIN_SIZE": 18,
    "PRIMER_MAX_SIZE": 25,
    "PRIMER_OPT_TM": 60.0,
    "PRIMER_MIN_TM": 57.0,
    "PRIMER_MAX_TM": 63.0,
    "PRIMER_MIN_GC": 40.0,
    "PRIMER_MAX_GC": 60.0,
    "PRIMER_PRODUCT_SIZE_RANGE": [[150, 300]]
}

r = primer3.bindings.design_primers(
    {"SEQUENCE_ID": "ACTB", "SEQUENCE_TEMPLATE": seq},
    settings
)

print("Primer pairs returned:", r["PRIMER_PAIR_NUM_RETURNED"])
```

```python



print("PRIMER PICKING RESULTS FOR NM_001101.5")


print("\nOLIGO")
print("LEFT PRIMER")
print("Start   :", r["PRIMER_LEFT_0"][0])
print("Length  :", r["PRIMER_LEFT_0"][1])
print("Tm      :", round(r["PRIMER_LEFT_0_TM"], 2), "°C")
print("GC%     :", round(r["PRIMER_LEFT_0_GC_PERCENT"], 2), "%")
print("Any TH  :", round(r["PRIMER_LEFT_0_SELF_ANY_TH"], 2))
print("3-prime :", round(r["PRIMER_LEFT_0_SELF_END_TH"], 2))
print("Hairpin :", round(r["PRIMER_LEFT_0_HAIRPIN_TH"], 2))
print("Sequence:", r["PRIMER_LEFT_0_SEQUENCE"])

print("\nRIGHT PRIMER")
print("Start   :", r["PRIMER_RIGHT_0"][0])
print("Length  :", r["PRIMER_RIGHT_0"][1])
print("Tm      :", round(r["PRIMER_RIGHT_0_TM"], 2), "°C")
print("GC%     :", round(r["PRIMER_RIGHT_0_GC_PERCENT"], 2), "%")
print("Any TH  :", round(r["PRIMER_RIGHT_0_SELF_ANY_TH"], 2))
print("3-prime :", round(r["PRIMER_RIGHT_0_SELF_END_TH"], 2))
print("Hairpin :", round(r["PRIMER_RIGHT_0_HAIRPIN_TH"], 2))
print("Sequence:", r["PRIMER_RIGHT_0_SEQUENCE"])

print("\nSEQUENCE SIZE :", len(seq), "bp")
print("PRODUCT SIZE  :", r["PRIMER_PAIR_0_PRODUCT_SIZE"], "bp")
print("PAIR ANY TH   :", r["PRIMER_PAIR_0_COMPL_ANY_TH"])
print("PAIR 3-PRIME  :", r["PRIMER_PAIR_0_COMPL_END_TH"])

```
