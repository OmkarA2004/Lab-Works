```python
import requests
from Bio import SeqIO
from io import StringIO
```

```python
accession = "BN000272"

print("TPA mRNA accession:", accession)
```

```python
url = "https://www.ebi.ac.uk/Tools/dbfetch"

params = {
    "db": "ena_sequence",
    "id": accession,
    "format": "fasta",
    "style": "raw"
}

print("Database:", params["db"])
print("Accession:", accession)
```

```python
response = requests.get(url, params=params)

print("Status code:", response.status_code)
```

```python
fasta_data = response.text

print(fasta_data)
```

```python
record = SeqIO.read(StringIO(fasta_data), "fasta")

print("Accession:", record.id)
print("Description:", record.description)
print("Sequence length:", len(record.seq))
```

```python
print("TPA mRNA sequence:")
print(record.seq)
```
